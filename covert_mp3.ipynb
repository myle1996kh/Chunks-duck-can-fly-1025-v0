```python
import os
from moviepy.editor import AudioFileClip

def convert_webm_to_mp3(input_folder: str, output_folder: str):
    """
    Chuyển tất cả file .webm trong input_folder sang .mp3 
    và lưu vào output_folder
    
    Args:
        input_folder (str): Thư mục chứa file .webm
        output_folder (str): Thư mục chứa file .mp3 đầu ra
    """
    # Tạo thư mục output nếu chưa có
    os.makedirs(output_folder, exist_ok=True)

    # Duyệt tất cả file trong thư mục input
    for filename in os.listdir(input_folder):
        if filename.lower().endswith(".webm"):
            input_path = os.path.join(input_folder, filename)
            output_filename = os.path.splitext(filename)[0] + ".mp3"
            output_path = os.path.join(output_folder, output_filename)

            try:
                # Load audio từ file webm
                clip = AudioFileClip(input_path)
                # Ghi ra file mp3
                clip.write_audiofile(output_path)
                clip.close()
                print(f"✅ Đã chuyển: {filename} -> {output_filename}")
            except Exception as e:
                print(f"❌ Lỗi với {filename}: {e}")
```

```python
import os
from pydub import AudioSegment

def convert_webm_to_mp3(input_folder: str, output_folder: str):
    """
    Chuyển tất cả file .webm trong input_folder sang .mp3 
    và lưu vào output_folder
    """
    os.makedirs(output_folder, exist_ok=True)

    for filename in os.listdir(input_folder):
        if filename.lower().endswith(".webm"):
            input_path = os.path.join(input_folder, filename)
            output_filename = os.path.splitext(filename)[0] + ".mp3"
            output_path = os.path.join(output_folder, output_filename)

            try:
                audio = AudioSegment.from_file(input_path, format="webm")
                audio.export(output_path, format="mp3")
                print(f"✅ Đã chuyển: {filename} -> {output_filename}")
            except Exception as e:
                print(f"❌ Lỗi với {filename}: {e}")
```

```python
# Ví dụ: input folder chứa file webm, output folder để chứa file mp3
convert_webm_to_mp3("record", "mp3_files")
```
